```python
if df["Fare"].isnull().sum()>0:
    df["Fare"]=df["Fare"].fillna(df['fare'].median())
    
```

```python
df['Fare'] = df['Fare'].round(2)
```

```python
df['Fare'] = df['Fare'].apply(lambda x: max(0.0, x))
```

**To check which special charcter ticket column have**

```python
import re

all_tickets = "".join(df['Ticket'].dropna().astype(str))

special_chars = set(re.findall(r'[^a-zA-Z0-9\s]', all_tickets))

print("Unique Special Characters Found in 'ticket_number':")
print(sorted(list(special_chars)))
```

```python
# Clean up dots and slashes, keeping spaces intact
df['Ticket'] = (
    df['Ticket']
    .astype(str)
    .str.replace('.', '', regex=False)
    .str.replace('/', '', regex=False)
    .str.strip()
)
```

```python
# 1. Summary of remaining missing values across ALL columns
print("=== MISSING VALUES SUMMARY ===")
print(df.isnull().sum())

# 2. Check current column names & data types
print("\n=== DATA TYPES ===")
print(df.dtypes)

# 3. Preview first 5 rows of your cleaned DataFrame
print("\n=== CLEANED DATASET PREVIEW ===")
display(df.head())
```

```python
median_age = df['Age'].median()

df['Age'] = df['Age'].fillna(median_age)

print("Remaining missing ages:", df['Age'].isnull().sum())
```

```python
# Drop the last column using its index position (-1)
df = df.drop(df.columns[-1], axis=1)

# Overwrite titanic_cleaned.csv to save changes permanently
df.to_csv('titanic_cleaned.csv', index=False)
```
