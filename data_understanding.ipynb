# Part 2: Feature Engineering


```python
info_df.head(10)
```

```python
info_df["Family_size"]=info_df["Sibling_Spouse"]+info_df["Parent_Children"]+1
```

```python
info_df
```

```python
#making column of age group
def age_group(age):
    if age<12:
        return "child"
    elif age<18:
        return "Teen"
    elif age<65:
        return "Adult"
    else:
        return "Old_Age"
info_df["Age_Category"]=info_df["Age"].apply(age_group)
#making column of fare category
def fare_distribution(fare):
    if fare<200:
        return "Low"
    elif fare>200 and fare<400:
        return "Medium"
    else:
        return "High"
info_df["Fare_Category"]=info_df["Fare"].apply(fare_distribution)
info_df
```
